```python
numeric_cols = Data.select_dtypes(include='number').columns

for col in numeric_cols:
    Data[col] = Data[col].fillna(Data[col].median())
```

```python
categorical_cols = Data.select_dtypes(include='object').columns

for col in categorical_cols:
    Data[col] = Data[col].fillna(Data[col].mode()[0])
```

```python
Data['Age_Group'] = pd.cut(
    Data['Age'],
    bins=[0, 18, 22, 26, 100],
    labels=['Teen', 'Young Adult', 'Adult', 'Older Adult']
)
```

```python
Data['AI_Usage_Level'] = pd.cut(
    Data['Daily_AI_Tool_Usage_Hours'],
    bins=[0, 2, 5, 8, float('inf')],
    labels=['Low', 'Moderate', 'High', 'Very High']
)
```

```python
Data['Social_Media_Usage_Level'] = pd.cut(
    Data['Daily_Social_Media_Hours'],
    bins=[0, 2, 5, 8, float('inf')],
    labels=['Low', 'Moderate', 'High', 'Very High']
)
```

```python
Data['Total_Digital_Usage_Hours'] = (
    Data['Daily_Social_Media_Hours'] +
    Data['Daily_AI_Tool_Usage_Hours']
)
```

```python
Data['AI_Social_Ratio'] = (
    Data['Daily_AI_Tool_Usage_Hours'] /
    (Data['Daily_Social_Media_Hours'] + 0.01)
)
```

```python
Data['Overall_Health_Score'] = (
    Data['Mental_Health_Score'] +
    Data['Physical_Health_Score']
) / 2
```

```python
Data['Activity_Sleep_Ratio'] = (
    Data['Physical_Activity_Hours'] /
    (Data['Sleep_Hours'] + 0.01)
)
```

##  **Machine Learning**

```python
# Features: Drop Student_ID, Mental_Health_Score (as it's the target), and other ID/non-predictive columns.
# Also remove newly created categorical features from X, as they will be created by the preprocessor.
X = Data.drop(columns=[
    'Student_ID',
    'Mental_Health_Score',
    'Sleep_Category' # This is a categorical feature derived from 'Sleep_Hours', we'll let OHE handle it.
])

# Target: Mental_Health_Score
y = Data['Mental_Health_Score']

print("X Shape:", X.shape)
print("y Shape:", y.shape)

print("\nTarget Distribution (Mental_Health_Score description):")
display(y.describe())
```

```python
# Identify numerical and categorical features for the updated X
numeric_features = X.select_dtypes(include=['number']).columns.tolist()
categorical_features = X.select_dtypes(include=['object', 'category']).columns.tolist()

print("Numerical Features (for X):")
print(numeric_features)

print("\nCategorical Features (for X):")
print(categorical_features)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np # Import numpy for sqrt in RMSE calculation
```

```python
numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(
        handle_unknown='ignore',
        drop='first' # Drop first to avoid multicollinearity
    ))
])
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_pipeline, numeric_features),
        ('cat', categorical_pipeline, categorical_features)
    ],
    remainder='passthrough' # Keep other columns if any, though none expected here
)
```

```python
print("Missing values in X before split:", X.isnull().sum().sum())
print("Missing values in y before split:", y.isnull().sum())
```

```python
# No missing values handling needed as we use SimpleImputer in the pipeline.
# This cell can be removed or kept as a check for any residual NaNs outside the pipeline features.
# For now, let's keep it and note that imputer in pipeline handles relevant missing data.
# Based on info(), there are no missing values in original Data, so `SimpleImputer` might not be strictly necessary
# but good for robustness.
```

```python
# Re-check the column lists after defining X and y in case any discrepancies exist
# For simplicity, we will use the full X and y as defined initially, assuming no NaNs for target
# and that preprocessor handles features in X.
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("X_train processed shape:", X_train_processed.shape)
print("X_test processed shape:", X_test_processed.shape)
```

```python
def predict_mental_health_score(student_data_raw):

    # Make a copy to avoid SettingWithCopyWarning if input is a view
    student_data = student_data_raw.copy()

    # -----------------------------
    # Feature Engineering (must match training data prep)
    # -----------------------------

    # 1. Total Daily Screen Exposure
    student_data['Total_Screen_Time'] = student_data['Daily_Social_Media_Hours'] + student_data['Daily_AI_Tool_Usage_Hours']

    # 2. Sleep Health Categorization
    # Ensure Sleep_Category is created using the same bins and labels
    student_data['Sleep_Category'] = pd.cut(
            student_data['Sleep_Hours'],
                bins=[0, 6, 8, 24],
                    labels=['Sleep Deprived (<6h)', 'Optimal (6-8h)', 'Extended (>8h)']
    )

    # 3. Screen-to-Active Ratio
    student_data['Screen_to_Active_Ratio'] = student_data['Total_Screen_Time'] / (student_data['Physical_Activity_Hours'] + 0.1)

    # Reindex columns to match the training data's feature order
    # The 'X' dataframe from training defines the expected feature set
    # Need to make sure the input has all required columns for `X`
    required_cols = list(X.columns)

    # Add missing columns with NaN values if they don't exist in student_data
    missing_cols = set(required_cols) - set(student_data.columns)
    for c in missing_cols:
        student_data[c] = np.nan # Imputer in pipeline will handle these

    student_data = student_data[required_cols]

    # Prediction
    prediction = model.predict(student_data)[0]

    return {
        'Predicted_Mental_Health_Score': round(prediction, 2)
    }
```
